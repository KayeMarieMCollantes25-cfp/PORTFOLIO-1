    # Program: Current Exchange (USD TO PHP)

##### Converting USD TO PHP alllows comparison of goods and cost of services in different country. This can be use for travelling, business trading, and investments for me and others.

```python
# Get the amount in USD
usd_amount = input("Enter amount in USD: $")
usd_amount = float(usd_amount)

# Find the exchange rate
exchange_rate = input("Enter exchange_rate in PHP: PHP")
exchange_rate = float(exchange_rate)

# all done
converted_amount = usd_amount * exchange_rate
print("Converted_amount: PHP", converted_amount)
```

    # Market Price

    # General Weighted Average

##### This code is used for evaluating the student's academic performance, to determine the elligibility for scholarships, and basis for honors.

```python
# Get the grade for each subjects
calculus = input("Enter calculus grade: ")
calculus = float(calculus)
anachem = input("Enter anachem grade: ")
anachem = float(anachem)
anachemlab = input("Enter anachemlab grade: ")
anachemlab = float(anachemlab)
comprog = input("Enter comprog grade: ")
comprog = float(comprog)
physics = input("Enter physics grade: ")
physics = float(physics)
physicslab = input("Enter physicslab grade: ")
physicslab = float(physicslab)
pathfit = input("Enter pathfit grade: ")
pathfit = float(pathfit)
artapp = input("Enter artapp grade: ")
artapp = float(artapp)
entrep = input("Enter entrep grade: ")
entrep = float(entrep)

# Calculation of Averafe
average = (calculus + anachem + anachemlab + comprog + physics + physicslab + pathfit + artapp + entrep) / 9

# all done
print("Your average grade is: ", average)
```
